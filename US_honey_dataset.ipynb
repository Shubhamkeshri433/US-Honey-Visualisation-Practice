## Which state has produced the highest amount of honey throughout all the available years?
    represent result using horizontal bar graph,
    or the prodduction values in desceinding order.

```python
High_prof_state = df.groupby('state')['production'].sum().sort_values(ascending=False).reset_index()
plt.figure(figsize=(12,8))
sns.barplot(data=High_prof_state, x='production', y='state', hue='state', palette='coolwarm')
plt.title('Honey Production by State')
plt.xlabel('Production')
plt.ylabel('State')
plt.grid(alpha=0.3)
plt.show()
```

# Which state has highest and lowest yeild per colony throughout all the years.

    represent result using h bar graph,
    group data based on state,
    calculate average yeild per colony,
    sort the values in descending order,
    visualize the result using a bar plot.

```python
yield_per_col = df.groupby('state')['yield_per_colony'].mean().sort_values(ascending=False).reset_index()
print(yield_per_col.head())

plt.figure(figsize=(12,8))
sns.barplot(data=yield_per_col, x='yield_per_colony', y='state', hue='state', palette='summer')
plt.title('Average Yield/Colony by State')
plt.xticks(rotation=90)
plt.show()
```

# Find which state had the highest honey stock in the year 2020.

    filter the dataset to include only the year 2020,
    group the filtered data based on state,
    calculate the total stocks for each state,
    sort the results in descending order,
    visualise the result using bar graph.

```python
df_2020 = df[df['year'] == 2020]
Gd = df_2020.groupby('state')['stocks'].sum().sort_values(ascending=False).reset_index()
plt.figure(figsize=(12,8))
sns.barplot(data=Gd, x='state', y='stocks', hue='state', palette='summer')
plt.title('Stock by State in 2020 Year')
plt.xticks(rotation = 90)
plt.show()
```

# Create a heatmap to visualize correlation between numerical columns in us honey dataset

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Heatmap')
plt.show()
```
